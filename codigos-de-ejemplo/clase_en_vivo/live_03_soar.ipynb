## 1. El Runbook (documento)

```python
RUNBOOK = """
RUNBOOK: Responder a fuerza bruta SSH
=====================================
PASO 1 - Recolectar info del incidente (IP, usuario, intentos)
PASO 2 - Verificar si la IP es conocida (CMDB / historial)
PASO 3 - Mitigar: bloquear IP en firewall y forzar cambio de password
PASO 4 - Documentar: crear ticket y notificar al equipo
"""
print(RUNBOOK)
```

## 2. El Playbook (código)
Los mismos 4 pasos, ejecutables. Con **scoring** y **human-in-the-loop** para lo riesgoso.

```python
def responder_fuerza_bruta(ip, usuario, intentos):
    print(f"\n[1] Incidente: {ip} atacó a '{usuario}' ({intentos} intentos)")

    # [2] ¿IP interna/conocida? -> probable falso positivo
    if ip.startswith("10."):
        return "Falso positivo: IP interna conocida"

    # Scoring simple de riesgo
    score = min(intentos * 10, 100)
    print(f"[2] Score de riesgo: {score}/100")

    # [3] Acciones automáticas de bajo riesgo
    print(f"[3] Bloqueando {ip} en firewall + forzando cambio de password")

    # human-in-the-loop para acciones de ALTO impacto
    if score >= 80:
        print("[!] Score alto -> requiere APROBACIÓN humana antes de aislar el host")

    # [4] Documentar
    print(f"[4] Ticket creado para {usuario}")
    return f"Gestionado (score {score})"

print(responder_fuerza_bruta("185.220.101.47", "admin", 9))
print(responder_fuerza_bruta("10.0.0.5", "juan", 3))
```
